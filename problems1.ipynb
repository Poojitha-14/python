```python
ifsc_code = input("Enter the IFSC code: ").upper()

is_valid = (
    len(ifsc_code) == 11 and
    ifsc_code[:4].isalpha() and
    ifsc_code[4] == '0' and
    ifsc_code[5:].isalnum()
)

if is_valid:
    print(f"The IFSC code '{ifsc_code}' is VALID.")
else:
    print(f"The IFSC code '{ifsc_code}' is INVALID.")
```
