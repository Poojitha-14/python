```python
num = 5
print(f"multiplication table of {num}:")
print("-----------------")
for j in range(1, 15): # Iterate beyond 10 to demonstrate 'break'
  if j % 2 != 0: # If j is odd, skip to the next iteration
    continue
  if j > 10: # If j exceeds 10, stop the loop
    break
  print(f"{num} x {j} = {num * j}")
```
