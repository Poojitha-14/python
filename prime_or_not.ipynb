```python
num=int(input("enter a number: "))
if num>1:#primes are starts 2,3,5,7,11..
  for i in range(2,num):#starts 2 to entered number
    if(num%i)==0:
      print(num,"is not a prime number")
      break
    i+=1
  else:
    print(num,"is a prime number")
else:
  print(num,"is not a prime number")
```
