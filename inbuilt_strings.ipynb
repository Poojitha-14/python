```python
print("abc".islower())#true
print("abc".isupper())
print("ABC".isupper())
print("ABC".islower())
print("pytHON".upper())#PYTHON
print("pytHON".lower())#python
```

```python
str="Python Strings"
for i in str:
  print(i)
```

```python
str="PytHON StRiNgS"
print(str.capitalize())#1st letter
print(str.title())#first letter of every character
print(str.casefold())#all lowercase
print(str.count('S'))
print(str.endswith('S'))
print(str.endswith('P'))
print(str.startswith('S'))
print(str.startswith('P'))
print(str.center(25,"*"))
print(str.encode())
print(str.endswith('N'))
print(str.expandtabs(8))
print(str.find('HON'))
print(str.)
```

```python
str="scott@oracle.com"
a=str.find('@')
d=str.find('.')
print(str[0:a])
print(str[a+1:d])
```

```python
'''
ord()--converts any character into ascii value
chr()--converts ascii(number) value into a character
'''

for ch in range(ord('A'),ord('Z')+1):
  print(chr(ch))
```
