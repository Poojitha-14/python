```python
size=input("enter the size of the pizza: ")
toppings_count=int(input("enter the number of toppings: "))
if size== "small":
  total_bill=10
elif size=="medium":
    total_bill=15
elif size=="large":
    total_bill=20
else:
  print("invalid pizza size: ")
for i in range(toppings_count):
  topping=input("enter your toppings : ")
  if topping=="cheese":
    total_bill+=2
  elif topping=="pepperoni":
    total_bill+=3
  elif topping=="olives":
    total_bill+=5
  elif topping=="jalapenos":
    total_bill+=5
  else:
    print("invalid topping")
print("total bill = $",total_bill)

```
