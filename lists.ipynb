```python
def circle_area(r):
  """Calculates the area of a circle given its radius."""
  area = 3.14159 * r**2
  return area
```

### Circle Area Function

The `circle_area(r)` function calculates the area of a circle. It takes one argument:

- `r`: The radius of the circle.

The function returns the calculated area using the formula `π * r^2`.

```python
# Example usage of the circle_area function
radius = 5
area = circle_area(radius)
print(f"The area of a circle with radius {radius} is: {area:.2f}")

radius_large = 10.5
area_large = circle_area(radius_large)
print(f"The area of a circle with radius {radius_large} is: {area_large:.2f}")
```
