```python
std={
    101:'scott',102:'pooji',103:'cherry',104:'james',105:'david'
}
print(std)
std[106]='xyz'
print(std)
del std[101]
del std[106]
print(std)
#check 101,104,105
print(101 in std)#false
print(104 in std)#true
print(105 in std)#true
```
