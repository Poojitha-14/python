```python
#parameterized constructor
class Product:
  def __init__(self,name,price,color,stock):
    print("product object is created...!")
    self.name=name
    self.price=price
    self.color=color
    self.stock=stock
    print("---------------")
p1=Product('phone',2000,'green',True)
print(f"name={p1.name}")
print(f"price={p1.price}")
print(f"color={p1.color}")
print(f"stock={p1.stock}")
```

```python
class Student():
  def __init__(self,name,age,id,branch):
    print("student object is created..")
    self.name=name
    self.age=age
    self.id=id
    self.branch=branch
def details(self):
  print("---------------")
  print(f"name is {self.name}")
  print(f"age is {self.age}")
  print(f"id is {self.id}")
  print(f"branch is {self.branch}")
s1=Student('scott',23,4208,'csm')
details(s1)
s2=Student('pooji',22,4532,'cai')
details(s2)
```
