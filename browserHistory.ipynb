```python
#stack
class BrowserHistory:
  def __init__(self,homepage):
    self.back   = [homepage]
    self.forward= []
  def visit(self,url):
    self.back.append(url)
    self.forward.clear()
  def go_back(self):
    if len(self.back) > 1:
      self.forward.append(self.back.pop())
      return self.back[-1]
  def go_forward(self):
    if self.forward:
      page=self.forward.pop()
      self.back.append(page)
      return page
#demo
b=BrowserHistory("google.com")
b.visit("youtubecom")      #back:[google,youtube]
b.visit("github.com")     #back:[google,youtube.github]
print(b.back)
print(b.forward)
print(b.go_back())
print(b.forward)
print(b.go_forward())
print(b.forward)
```
