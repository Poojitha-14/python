```python
class Mobile():
  def __init__(self,brand,price,color,stock,guarranty):
    print("mobile objects created...")
    self.brand=brand
    self.price=price
    self.color=color
    self.stock=stock
    self.guarranty=guarranty
    print("---------------")
p1=Mobile('mobile phone',20000,'green','True','5 years')
print(f"brand={p1.brand}")
print(f"price={p1.price}")
print(f"color={p1.color}")
print(f"stock={p1.stock}")
print(f"guarranty={p1.guarranty}")
```
